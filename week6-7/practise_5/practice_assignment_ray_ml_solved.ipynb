
# Exercise 6: Detect Outliers

## Background


Management suspects unusually high bills.


## Dataset / Input


monthly_bill column


## Task


Identify outliers using IQR.


## Hint


Use quantiles.



```python

# =====================================================
# TODO
# =====================================================


# Detect outliers

```


## Solution


```python

q1 = df["monthly_bill"].quantile(0.25)
q3 = df["monthly_bill"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = df[
    (df["monthly_bill"] < lower)
    |
    (df["monthly_bill"] > upper)
]

outliers
```


# Exercise 18: Using ray.wait

## Background


Some tasks finish before others.


## Dataset / Input


Tasks


## Task


Wait for first completed task.


## Hint


Use ray.wait().



```python

# =====================================================
# TODO
# =====================================================


# Wait for tasks

```


## Solution


```python

import time

@ray.remote
def work(x):

    time.sleep(x)

    return x

tasks = [

    work.remote(i)

    for i in [3,1,2]

]

ready, pending = ray.wait(
    tasks,
    num_returns=1
)

print(ray.get(ready))
```


# Exercise 19: Create Actor

## Background


Some applications require state.


## Dataset / Input


Counter Example


## Task


Create an Actor.


## Hint


Use class + @ray.remote.



```python

# =====================================================
# TODO
# =====================================================


# Create actor

```


## Solution


```python

@ray.remote
class Counter:

    def __init__(self):

        self.count = 0

    def increment(self):

        self.count += 1

        return self.count

counter = Counter.remote()

print(
    ray.get(
        counter.increment.remote()
    )
)
```


# Exercise 20: Stateful Actor

## Background


Actors maintain state between calls.


## Dataset / Input


Counter


## Task


Increment counter multiple times.


## Hint


Call actor methods repeatedly.



```python

# =====================================================
# TODO
# =====================================================


# Increment actor

```


## Solution


```python

@ray.remote
class Counter:

    def __init__(self):

        self.count = 0

    def increment(self):

        self.count += 1

        return self.count

counter = Counter.remote()

for _ in range(5):

    print(
        ray.get(
            counter.increment.remote()
        )
    )
```


# Exercise 37: Mini Project: Hyperparameter Search

## Background


You are training a churn
prediction model.

Management wants the best
hyperparameters.


## Dataset / Input


Search Space

learning_rate

0.001
0.01
0.1


## Task


Perform a Ray Tune search
and identify the best result.


## Hint


Use grid_search().



```python

# =====================================================
# TODO
# =====================================================


# Complete tuning project
```


## Solution


```python

from ray import tune

def objective(config):

    score = (

        config["lr"] - 0.01

    ) ** 2

    return {

        "score":score

    }

tuner = tune.Tuner(

    objective,

    param_space={

        "lr":

        tune.grid_search(

            [
                0.001,
                0.01,
                0.1
            ]

        )

    }

)

results = tuner.fit()

print(

    results.get_best_result(

        metric="score",

        mode="min"

    ).config

)
```
